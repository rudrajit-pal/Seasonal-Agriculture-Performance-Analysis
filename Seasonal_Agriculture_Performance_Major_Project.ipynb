```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_excel("/content/seasonal_agriculture_performance_dataset.xlsx")
```

```python
# Check the data types and non-null values
df.info()
```

```python
# Generate a statistical summary
df.describe()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Fill missing numerical values with the median
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())
df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())
```

```python
# Verify missing values after treatment
df.isnull().sum()
```

```python
# Check duplicate records
df.duplicated().sum()
```

```python
# Plot histograms of numerical variables
df.hist(figsize=(15, 12))
plt.tight_layout()
plt.show()
```

```python
# Check the distribution of seasons
df["Season"].value_counts()
```

```python
# Check the distribution of crops
df["Crop"].value_counts()
```

```python
# Check the distribution of irrigation methods
df["Irrigation_Method"].value_counts()
```

```python
# Check the distribution of states
df["State"].value_counts()
```

```python
# Compare average production, revenue, cost and profit across seasons
df.groupby("Season")[["Production_Tonnes", "Revenue_INR", "Total_Cost_INR", "Profit_INR"]].mean()
```

```python
# Compare average environmental conditions across seasons
df.groupby("Season")[["Rainfall_mm", "Avg_Temperature_C", "Humidity_pct", "Sunlight_Hours_Day"]].mean()
```

```python
# Visualize environmental conditions across seasons
df.groupby("Season")[["Rainfall_mm", "Avg_Temperature_C", "Humidity_pct", "Sunlight_Hours_Day"]].mean().plot(kind="bar", figsize=(10, 10))

plt.title("Average Environmental Conditions by Season")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

```python
# Compare average resource usage across seasons
df.groupby("Season")[["Fertilizer_kg_ha", "Pesticide_Litre_ha", "Water_Used_m3"]].mean()
```

```python
# Compare average soil conditions across seasons
df.groupby("Season")[["Soil_pH", "Soil_Moisture_pct"]].mean()
```

```python
# Visualize soil conditions across seasons
df.groupby("Season")[["Soil_pH", "Soil_Moisture_pct"]].mean().plot(kind="bar", figsize=(8, 5))

plt.title("Average Soil Conditions by Season")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.show()
```

```python
# Calculate correlations between environmental factors and yield
df[["Rainfall_mm", "Avg_Temperature_C", "Humidity_pct", "Sunlight_Hours_Day", "Yield_Tonnes_Ha"]].corr()["Yield_Tonnes_Ha"].sort_values(ascending=False)
```

```python
# Compare average crop yield by season and crop
df.groupby(["Season", "Crop"])["Yield_Tonnes_Ha"].mean().unstack()
```

```python
# Compare average profit by season and crop
df.groupby(["Season", "Crop"])["Profit_INR"].mean().unstack()
```

```python
# Compare average water efficiency across seasons
df.groupby("Season")["Water_Efficiency_t_per_1000m3"].mean()
```

```python
# Compare average disease and pest risk across seasons
df.groupby("Season")["Disease_Pest_Risk_pct"].mean()
```

```python
# Calculate correlations between important agricultural variables
correlation_matrix = df[[
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Water_Used_m3",
    "Revenue_INR",
    "Profit_INR",
    "Disease_Pest_Risk_pct"
]].corr()

correlation_matrix
```

```python
# Visualize correlations between agricultural variables
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix of Agricultural Variables")
plt.show()
```

```python
# Compare average yield by irrigation method
df.groupby("Irrigation_Method")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)
```

```python
# Plot the distribution of crop yield
sns.histplot(data=df, x="Yield_Tonnes_Ha", kde=True)

plt.title("Distribution of Crop Yield")
plt.xlabel("Yield (Tonnes/Ha)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Plot the distribution and outliers of crop yield
sns.boxplot(y=df["Yield_Tonnes_Ha"])

plt.title("Boxplot of Crop Yield")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()
```

```python
# Compare yield distribution across seasons
sns.boxplot(data=df, x="Season", y="Yield_Tonnes_Ha")

plt.title("Yield Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()
```

```python
# Plot the number of records for each season
sns.countplot(data=df, x="Season")

plt.title("Number of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")
plt.show()
```

```python
# Compare average yield across seasons
sns.barplot(data=df, x="Season", y="Yield_Tonnes_Ha")

plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.show()
```

```python
# Analyze the relationship between rainfall and crop yield
sns.scatterplot(data=df, x="Rainfall_mm", y="Yield_Tonnes_Ha", hue="Season")

plt.title("Rainfall vs Crop Yield by Season")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()
```

```python
# Compare important performance variables
performance_cols = ["Yield_Tonnes_Ha", "Production_Tonnes", "Revenue_INR", "Profit_INR"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, column in zip(axes.flatten(), performance_cols):
    sns.boxplot(y=df[column], ax=ax)
    ax.set_title(f"Boxplot of {column}")
    ax.set_ylabel(column)

plt.tight_layout()
plt.show()
```

```python
# Compare average profit across seasons
sns.barplot(data=df, x="Season", y="Profit_INR")

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.show()
```

```python
# Compare resource usage across seasons
resource_cols = ["Fertilizer_kg_ha", "Pesticide_Litre_ha", "Water_Used_m3"]

df.groupby("Season")[resource_cols].mean().plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Resource Usage by Season")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

```python
# Show the distribution of irrigation methods
df["Irrigation_Method"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7)
)

plt.title("Distribution of Irrigation Methods")
plt.ylabel("")
plt.show()
```
